```python
reference5_file_path = "../notebooks/sil_folder/reference/ours_binning_5.npz"

reference5 = np.load(reference5_file_path)

binning_results = reference5['binning_results']
labels_bin = reference5['true_labels']
covs = reference5['covs']

print(covs.shape)
print(reference5)
```

```python
threshold = 0.9

ratios_list = np.arange(0, 0.1, 0.01)

from evaluation.utils import align_labels_via_hungarian_algorithm
from sklearn.metrics import recall_score, precision_score, f1_score

cleaned_pred_labels = binning_results[binning_results != -1]
cleaned_true_labels = labels_bin[binning_results != -1]
cleaned_covs = covs[binning_results != -1]

print(cleaned_pred_labels.shape, cleaned_true_labels.shape, cleaned_covs.shape)

print(f"Unique labels: {len(set(pred_labels))}, Unique true labels: {len(set(cleaned_true_labels))}")

# Align labels
alignment_bin = align_labels_via_hungarian_algorithm(cleaned_true_labels, cleaned_pred_labels)
aligned_pred_labels = [alignment_bin.get(label) for label in cleaned_pred_labels]
hq_recall_bin_num_list, random_recall_hq_bin_list = [], []
hq_precision_bin_num_list, random_precision_hq_bin_list = [], []

random_strategy_recalls, our_strategy_recalls, hq_f1_bin_num_list = [], [], []
random_f1_hq_bin_list = []
for removal_ratio in ratios_list:

    num_remove = int(len(aligned_pred_labels) * removal_ratio)

    new_pred_labels = np.asarray(aligned_pred_labels).copy()
    new_true_labels = np.asarray(cleaned_true_labels).copy()
    new_covs = np.asarray(cleaned_covs).copy()
    # new_true_labels[idx] = -1

    # Random removal strategy
    idx = np.random.choice(len(aligned_pred_labels), size=num_remove, replace=False)
    new_pred_labels[idx] = -1

    new_labels = np.sort(np.unique(np.concatenate((new_true_labels, new_pred_labels))))
    # Count the occurrences of labels in labels inside true_labels
    weights = np.array([np.sum(new_true_labels == label) for label in new_labels])

    # raise ValueError(new_true_labels.shape, new_pred_labels.shape, new_labels.shape)
    all_recall = recall_score(y_true=new_true_labels, y_pred=new_pred_labels, labels=new_labels, average=None)
    all_precision = precision_score(y_true=new_true_labels, y_pred=new_pred_labels, labels=new_labels, average=None)
    all_f1 = f1_score(y_true=new_true_labels, y_pred=new_pred_labels, labels=new_labels, average=None)

    weighted_avg_recall = np.average(all_recall[1:], weights=weights[1:]) # Exclude the -1 label for recall calculation
    weighted_avg_precision = np.average(all_precision[1:], weights=weights[1:]) # Exclude the -1 label for precision calculation

    print(f"After removing {removal_ratio*100:.0f}% samples:")
    print(f"\t+ Random strategy: Recall: {weighted_avg_recall:.4f} | Precision: {weighted_avg_precision:.4f}")
    random_strategy_recalls.append(weighted_avg_recall)
    random_recall_hq_bin_list.append(sum(all_recall[1:] >= threshold))
    random_precision_hq_bin_list.append(sum(all_precision[1:] >= threshold))
    random_f1_hq_bin_list.append(sum(all_f1[1:] >= threshold))

    # Our strategy: remove the sequences having the smallest covariance determinant
    cov_dets = np.log(new_covs + 1).sum(axis=-1)  #np.sort(new_covs, axis=-1)[:, int(new_covs.shape[1]*0.5)] #np.log(new_covs + 1).sum(axis=-1)
    remove_indices = np.argsort(cov_dets)[::-1][:num_remove]
    # raise ValueError("dim,", chosen_indices.shape)

    new_pred_labels = np.asarray(aligned_pred_labels).copy()
    new_pred_labels[remove_indices] = -1

    new_labels = np.sort(np.unique(np.concatenate((new_true_labels, new_pred_labels))))
    # Count the occurrences of labels in labels inside true_labels
    weights = np.array([np.sum(new_true_labels == label) for label in new_labels])

    all_recall = recall_score(y_true=new_true_labels, y_pred=new_pred_labels, labels=new_labels, average=None)
    all_precision = precision_score(y_true=new_true_labels, y_pred=new_pred_labels, labels=new_labels, average=None)

    weighted_avg_recall = np.average(all_recall[1:], weights=weights[1:]) # Exclude the -1 label for recall calculation
    hq_recall_bin_num = sum(all_recall[1:] >= threshold)
    hq_recall_bin_num_list.append(hq_recall_bin_num)
    weighted_avg_precision = np.average(all_precision[1:], weights=weights[1:]) # Exclude the -1 label for precision calculation
    hq_precision_bin_num_list.append(sum(all_precision[1:] >= threshold))


    all_f1 = f1_score(y_true=new_true_labels, y_pred=new_pred_labels, labels=new_labels, average=None)
    hq_f1_bin_num_list.append(sum(all_f1[1:] >= threshold))

    # print(f"\t+ Our strategy: Recall: {weighted_avg_recall:.4f} | Precision: {weighted_avg_precision:.4f}")
    # our_strategy_recalls.append(weighted_avg_recall)

import matplotlib.pyplot as plt
# plt.figure()
# plt.plot(ratios_list*100, random_strategy_recalls, '-',  label='Random Removal Strategy')
# plt.plot(ratios_list*100, our_strategy_recalls, '--', label='Covariance-based Removal Strategy')
# plt.xlabel('Percentage of Removed Sequences (%)')
# plt.ylabel('Weighted Average Recall')
# plt.title('Comparison of Removal Strategies on Binning Performance')
# plt.legend()
# plt.grid(True)
# plt.show()
# # plt.savefig('../notebooks/removing_strategy_comparison.png')


print("HQ:", hq_recall_bin_num_list)
print("Random HQ:", random_recall_hq_bin_list)

plt.figure()
plt.plot(ratios_list*100, random_recall_hq_bin_list, '-',  label='Random Removal Strategy')
plt.plot(ratios_list*100, hq_recall_bin_num_list, '--', label='Covariance-based Removal Strategy')
plt.xlabel('Percentage of Removed Sequences (%)')
plt.ylabel('Number of HQ Bins (Recall >= 0.9)')
plt.title('Comparison of Removal Strategies on Number of HQ Bins')
plt.legend()
plt.grid(True)
plt.show()
#
plt.figure()
plt.plot(ratios_list*100, random_precision_hq_bin_list, '-',  label='Random Removal Strategy')
plt.plot(ratios_list*100, hq_precision_bin_num_list, '--', label='Covariance-based Removal Strategy')
plt.xlabel('Percentage of Removed Sequences (%)')
plt.ylabel('Number of HQ Bins (Precision >= 0.9)')
plt.title('Comparison of Removal Strategies on Number of HQ Bins')
plt.legend()
plt.grid(True)
plt.show()

plt.figure()
plt.plot(ratios_list*100, random_f1_hq_bin_list, '-',  label='Random Removal Strategy')
plt.plot(ratios_list*100, hq_f1_bin_num_list, '-',  label='Covariance Strategy')
plt.legend()
plt.grid(True)
plt.show()


# # print(recall)
# labels = np.sort(np.unique(np.concatenate((true_labels, aligned_pred_labels))))
# recall = recall_score(y_true=true_labels, y_pred=aligned_pred_labels, labels=labels, average=None)
#
# print(
#     recall_score(true_labels, aligned_pred_labels, average='weighted'),
# )
# # Count the occurrences of labels in labels inside true_labels
# num_true_labels = np.array([np.sum(true_labels == label) for label in labels])
#
# weighted_avg_recall = np.average(recall, weights=num_true_labels)
# print(f"Weighted Average Recall: {weighted_avg_recall}")


```
