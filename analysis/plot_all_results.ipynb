```python



names = [
    r"\textsc{Kmer(cosine)}", r"\textsc{Kmer($l_1$)}", r"\textsc{HyenaDNA}",  r"\textsc{DNABERT-2}",
    r"\textsc{DNABERT-S}", r"\textsc{RevisitKmers}", r"\textsc{UncertainGen}"
]
colors = ["tab:blue", "tab:cyan", "tab:olive", "tab:orange", "tab:red"][::-1]

# Plot 2 x 3 figures
fig, ax = plt.subplots(2, 3, figsize=(20, 12))

title = r"\textsl{Synthetic 5}"
data = np.asarray(synthetic5)
x = np.arange(data.shape[0])
yticks = synthetic5_xticks
for i in range(data.shape[1]):
    ax[0, 0].bar(x, data[:, i], align='center', label=names[i], alpha=alpha, color=colors[i], linewidth=linewidth, edgecolor=edgecolor)
ax[0, 0].set_title(title, fontsize=fontsize)
ax[0, 0].set_ylabel('Species count', fontsize=fontsize)
ax[0, 0].set_yticklabels(yticks, fontsize=fontsize)
ax[0, 0].set_xticks(x, [])
ax[0, 0].grid(axis='x')

title = r"\textsl{Plant 5}"
data = np.asarray(plant5)
yticks = plant5_xticks
x = np.arange(data.shape[0])
for i in range(data.shape[1]):
    ax[0, 1].bar(x, data[:, i], align='center', label=names[i], alpha=alpha, color=colors[i], linewidth=linewidth, edgecolor=edgecolor)
ax[0, 1].set_title(title, fontsize=fontsize)
ax[0, 1].set_yticklabels(yticks, fontsize=fontsize)
ax[0, 1].set_xticks(x, [])
ax[0, 1].grid(axis='x')

title = r"\textsl{Marine 5}"
data = np.asarray(marine5)
yticks = marine5_xticks
x = np.arange(data.shape[0])
for i in range(data.shape[1]):
    ax[0, 2].bar(x, data[:, i], align='center', label=names[i], alpha=alpha, color=colors[i], linewidth=linewidth, edgecolor=edgecolor)
ax[0, 2].set_title(title, fontsize=fontsize)
ax[0, 2].set_yticklabels(yticks, fontsize=fontsize)
ax[0, 2].set_xticks(x, [])
ax[0, 2].grid(axis='x')

title = r"\textsl{Synthetic 6}"
data = np.asarray(synthetic6)
x = np.arange(data.shape[0])
yticks = synthetic6_xticks
for i in range(data.shape[1]):
    ax[1, 0].bar(x, data[:, i], align='center', label=names[i], alpha=alpha, color=colors[i], linewidth=linewidth, edgecolor=edgecolor)
ax[1, 0].set_title(title, fontsize=fontsize)
ax[1, 0].set_ylabel('Species count', fontsize=fontsize)
ax[1, 0].set_yticklabels(yticks, fontsize=fontsize)
ax[1, 0].set_xticks(x+0.25, names, fontsize=fontsize, rotation=90, ha='right')
ax[1, 0].grid(axis='x')

title = r"\textsl{Plant 6}"
data = np.asarray(plant6)
yticks = plant6_xticks
x = np.arange(data.shape[0])
for i in range(data.shape[1]):
    ax[1, 1].bar(x, data[:, i], align='center', label=names[i], alpha=alpha, color=colors[i], linewidth=linewidth, edgecolor=edgecolor)
ax[1, 1].set_title(title, fontsize=fontsize)
ax[1, 1].set_yticklabels(yticks, fontsize=fontsize)
ax[1, 1].set_xticks(x+0.25, names, fontsize=fontsize, rotation=90, ha='right')
ax[1, 1].grid(axis='x')

title = r"\textsl{Marine 6}"
data = np.asarray(marine6)
yticks = marine6_xticks
x = np.arange(data.shape[0])
for i in range(data.shape[1]):
    ax[1, 2].bar(x, data[:, i], align='center', label=names[i], alpha=alpha, color=colors[i], linewidth=linewidth, edgecolor=edgecolor)
ax[1, 2].set_title(title, fontsize=fontsize)
ax[1, 2].set_yticklabels(yticks, fontsize=fontsize)
ax[1, 2].set_xticks(x+0.25, names, fontsize=fontsize, rotation=90, ha='right')
ax[1, 2].grid(axis='x')
print(names)


# Add the legend out of the plot and below in one line
legend = fig.legend(
    [r"$(0.5, \ 0.6]$", r"$(0.6, \ 0.7]$", r"$(0.7, \ 0.8]$", r"$(0.8, \ 0.9]$", r"$(0.9, \ 1.0]$"],
    loc='upper center', bbox_to_anchor=(0.5, 0.05), ncol=5, fontsize=fontsize
)
# Add legend title
legend.set_title("$F_1$ Score Ranges", prop={'size': fontsize})
plt.tight_layout(w_pad=0., h_pad=2.5, pad=4)
# plt.show()
plt.savefig(os.path.join(os.getcwd(), f'../paper/figures/all_results.png'), dpi=300, bbox_inches='tight')
```
